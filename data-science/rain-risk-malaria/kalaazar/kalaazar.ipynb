```python
import plotly.graph_objects as go
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
kalaazar_df = pd.read_csv("kalaazar-cases.csv")
T_Cases = []
for i in range(len(kalaazar_df)):
    T_Cases.append(sum(kalaazar_df.iloc[i,1:]))

print(T_Cases)
```

```python
import plotly.graph_objects as go

States = list(kalaazar_df['States'])[0:-1]

fig = go.Figure()
fig.add_trace(go.Bar(
    x=States,
    y=np.log2(T_Cases),
    name='Primary Product',
    marker_color='indianred'
))

# Here we modify the tickangle of the xaxis, resulting in rotated labels.
fig.update_layout(barmode='group', xaxis_tickangle=-45)
fig.show()
```
